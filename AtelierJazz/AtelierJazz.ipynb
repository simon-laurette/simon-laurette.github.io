# Atelier Jazz - Ressources

Bienvenue sur la page regroupant les ressources concernant les Ateliers Jazz de la Maison de la Musique
Vous y trouverez :
- l'historique des atelier et les documents/vidéos relatifs à chaque cours
- les documents théoriques généraux et récapitulatifs
