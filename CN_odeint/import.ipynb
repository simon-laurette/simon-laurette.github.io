# Importer des données d'un tableur vers Python sur Labnbook

> **Objectif** :
>
> - Récupérer sous Python des données rentrées dans un tableur Labnbook
 



### Problématique

Vous disposez sous Labnbook d'un tableur de relevés, associé à un graphique et vous souhaiteriez récupérer les données sous forme de tableaux Python afin de faire des traitements plus avancés.

<img src="https://simon-laurette.github.io/CN_odeint/images/tableur.png" width=90%>


### 1 - Exportation en fichier csv

Dans un premier temps, il faut exporter les données dans un fichier .csv 

Pour cela, cliquer sur le menu représenté ci-dessous et bien paramétrer le menu qui apparaît ensuite en précisant que **le séparateur décimal doit être un point.**


<img src="https://simon-laurette.github.io/CN_odeint/images/export1.png">


### 2 - Imporation du fichier csv sur Python

Dans un premier temps, il faut importer le fichier csv qui a été téléchargé sur votre PC dans les fichiers Ressources du Labnbook. Sélectionner ainsi le fichier grâce au menu ci-dessous et valider (ne pas oublier de lui donner un nom, ici ce sera "test").


<img src="https://simon-laurette.github.io/CN_odeint/images/import.png">

Ouvrir ensuite un labdoc de code et cliquer sur la petite icône "importer".


<img src="https://simon-laurette.github.io/CN_odeint/images/import2.png">


Un code apparaît spontanément et toutes les données du fichier sont stockées dans un objet appelé `data`, que vous pouvez visualiser comme ci-dessous :

<img src="https://simon-laurette.github.io/CN_odeint/images/import3.png">

Il est ensuite possible de récupérer les colonnes que vous voulez sous forme d'un tableau grâce à la syntaxe :

> `x=data.nomcolonne`

On récupére alors les colonnes que l'on souhaite et on peut effectuer le tracé :
